# """ While reconstructing the corrected order date, a few invalid date combinations (e.g., 2015-02-31) were detected due to mismatched day and month values. when I saw some row i.e. 943,944,945, we detect there is an error in the month column along with the year. 
# I noticed that in some rows the ship date month is less than the correct_order_date. 
# I noticed that one ship date was 2015-03-06 and the correct_order_date was 2015-05-29. How is it possible that the order can ship before it is placed? 

# To resolve this logically, we can take MONTH for correct_order_date from SHIP DATE and a custom validation function was applied that automatically
# adjusts any invalid day to the last valid day of that specific month (e.g., February → 28/29, April → 30).

# This approach preserves data consistency and eliminates conversion errors without distorting the temporal sequence."""

```python
import pandas as pd
import calendar


# Step 1️⃣: Combine Order Year, Ship Month, and Order Day into a single string
# Purpose: Create a corrected order date format by merging the year (from order_id_year),
# month (from Ship Date), and day (from Order Date). This helps standardize inconsistent month/year patterns.
df['Corr_Order_date_v2_str'] = (
    df['order_id_year'].astype(str) + "-" +
    df['Ship Date'].dt.month.astype(str).str.zfill(2) + "-" +
    df['Order Date'].dt.day.astype(str).str.zfill(2)
)


# Step 2️⃣: Define a function to fix invalid dates (e.g., 2015-02-31)
# Purpose: If a day value exceeds the valid limit for that month,
# it automatically replaces it with the month's last valid day.
def fix_invalid_date(date_str):
    try:
        return pd.to_datetime(date_str)
    except ValueError:
        year, month, day = map(int, date_str.split('-'))
        last_day = calendar.monthrange(year, month)[1]
        return pd.Timestamp(f"{year}-{month:02d}-{last_day:02d}")


# Step 3️⃣: Apply the fix function to create a corrected date column
df['Corr_Order_date_v2'] = df['Corr_Order_date_v2_str'].apply(fix_invalid_date)


# Step 4️⃣: Adjust logical inconsistencies between Order and Ship Dates
# Purpose: In some records, the corrected Order Date still appears after the Ship Date
# (e.g., Order = 2015-03-29, Ship = 2015-03-06), which is illogical in real business terms.
# To maintain logical order and data continuity, such cases are corrected by
# setting the Order Date as one day before the Ship Date.
mask_invalid = df['Corr_Order_date_v2'] > df['Ship Date']
df.loc[mask_invalid, 'Corr_Order_date_v2'] = df.loc[mask_invalid, 'Ship Date'] - pd.Timedelta(days=1)


# Step 5️⃣: Drop the temporary string column
df.drop(columns='Corr_Order_date_v2_str', inplace=True)

# ✅ Final Output:
# The column 'Corr_Order_date_v2' now contains fully valid and logically consistent Order Dates.
# All invalid or unrealistic dates have been corrected automatically while preserving business meaning.



#Re-verify where there was a month/year error in the above row
df.loc[[943, 944, 945], ['order_id_year','Order Date','Corr_Order_date','Corr_Order_date_v2','Ship Date']]
```

# Analysis Report: Outlier Identification and Initial Cleaning of 'Sales Price'
# In the initial analysis of the 'Sales Price' column, a boxplot and a statistical summary were generated to understand the 
# data's distribution. The analysis clearly revealed that the data is heavily right-skewed, with the majority of sales concentrated
# at lower values and a significant number of high-value transactions identified as outliers. The most critical finding was a significant data error: 
# the minimum sales price was -31.5, which is a practical impossibility. This value is considered a data entry error rather than a statistical outlier.
# Therefore, as a crucial first step to ensure data integrity, all records containing negative sales prices have been removed. 
# This initial cleaning process provides a valid and reliable foundation for any further outlier handling and analysis.

```python
# 2. Create a function called remove_outliers which takes as parameter the dataframe and the column which needs to be searched for outliers. 
# Using the 3* IQR rule, the function should detect and remove the outliers to return the cleaned dataframe

def remove_outliers(df, column_name):
    # Calculate quartiles
    Q1 = df[column_name].quantile(0.25)
    Q3 = df[column_name].quantile(0.75)
    
    # Calculate IQR
    IQR = Q3 - Q1
    
    # Calculate bounds using the 3 * IQR rule
    upper_bound = Q3 + (3 * IQR)
    lower_bound = Q1 - (3 * IQR)
    
    # Identify and count outliers (optional but good practice)
    outliers = df[(df[column_name] < lower_bound) | (df[column_name] > upper_bound)]
    print(f"Number of outliers detected in '{column_name}': {len(outliers)}")
    
    # Remove outliers
    filtered_df = df[(df[column_name] >= lower_bound) & 
                     (df[column_name] <= upper_bound)]
    
    return filtered_df

# Usage with your DataFrame
# First, ensure you have handled the negative price error from Question 1
df_positive = df[df['Sales Price'] >= 0]

# Now, apply the outlier removal function
cleaned_df = remove_outliers(df_positive, 'Sales Price')

# Check the shape of the new DataFrame
print(f"\nOriginal DataFrame rows (positive values only): {len(df_positive)}")
print(f"Cleaned DataFrame rows: {len(cleaned_df)}")

# Display the head of the cleaned DataFrame
cleaned_df.head()
```

```python
# 3. Yearly Growth in Sales and Profit: Use a year-over-year growth chart to compare the sales and profit growth over different years.
import matplotlib.pyplot as plt

# --- Step 1: Your first step is perfect ---
# Create the summary of yearly totals.
df1 = df.groupby(df['Corr_Order_date_v2'].dt.year)[['Total Sales', 'Total Profit']].sum()

# --- Step 2: THIS IS WHERE YOU USE .pct_change() ---
# Calculate the percentage growth from the previous year and store it in a new variable.
yearly_growth = df1.pct_change() * 100 # Multiply by 100 to get a percentage

# --- Step 3: Now, plot the NEW 'yearly_growth' DataFrame ---
# A bar chart is usually better for showing growth.
yearly_growth.plot(kind='bar', figsize=(10,6))

# Add titles and labels
plt.title('Year-over-Year Growth in Sales and Profit', fontsize=14, fontweight='bold')
plt.xlabel('Year')
plt.ylabel('Growth (%)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.axhline(y=0, color='black', linewidth=0.8) # Add a zero line
plt.legend()
plt.tight_layout()

plt.show()
```
